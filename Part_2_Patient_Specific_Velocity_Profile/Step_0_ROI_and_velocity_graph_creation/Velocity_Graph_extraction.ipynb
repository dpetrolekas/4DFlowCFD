```python
# velocity profile visualization (v-component) (Optional)
import h5py
import numpy as np
import matplotlib.pyplot as plt
from scipy.interpolate import griddata

# Function to create the elliptical mask
def create_ellipse_mask(image_shape, center, axes, angle):
    x, y = np.meshgrid(np.arange(image_shape[1]), np.arange(image_shape[0]))  # Use image_shape
    yc, xc = center
    a, b = axes
    cos_angle = np.cos(np.radians(angle))
    sin_angle = np.sin(np.radians(angle))
    ellipse = (((x - xc) * cos_angle + (y - yc) * sin_angle) ** 2) / a ** 2 + \
              (((x - xc) * sin_angle - (y - yc) * cos_angle) ** 2) / b ** 2
    mask = ellipse <= 1
    return mask

# Function to visualize the elliptical ROI as a continuous shape for all timesteps
def visualize_ellipse_for_all_timesteps(h5_file_path):
    # Load the HDF5 file
    with h5py.File(h5_file_path, "r") as f:
        v = f["v"][:]  # Assuming the dataset key is 'v'

    # Define ellipse parameters
    center = (35, 51)  # Adjusted center (y, x)
    axes = (10, 6.5)  # (semi-major axis, semi-minor axis)
    angle = 0  # Rotation angle in degrees (ellipse is vertical)

    # Iterate over all timesteps
    num_timesteps = v.shape[0]
    for timestep in range(num_timesteps):
        # Select the current slice to analyze
        slice_v = v[timestep, :, :, :]  # Current timestep
        slice_57_v = slice_v[:, 57, :]  # Extract slice 57 from the third dimension

        # Calculate elliptical mask
        ellipse_mask = create_ellipse_mask(slice_57_v.shape, center, axes, angle)

        # Extract ROI values
        x_coords, y_coords = np.meshgrid(
            np.arange(slice_57_v.shape[1]), np.arange(slice_57_v.shape[0])
        )
        x_roi = x_coords[ellipse_mask]
        y_roi = y_coords[ellipse_mask]
        values_roi = slice_57_v[ellipse_mask]

        # Interpolate to create a smooth grid within the ellipse
        grid_x, grid_y = np.meshgrid(
            np.linspace(38, 70, 300),  # Interpolation range for x-axis
            np.linspace(10, 60, 300),  # Interpolation range for y-axis
        )
        grid_values = griddata(
            (x_roi, y_roi), values_roi, (grid_x, grid_y), method="cubic", fill_value=np.nan
        )

        # Plot the interpolated ROI as a continuous shape
        plt.figure(figsize=(10, 8))
        contour = plt.contourf(grid_x, grid_y, grid_values, levels=9, cmap="viridis")
        plt.colorbar(contour, label="Velocity (m/s)")
        plt.title(f"Velocity Distribution at Timestep {timestep}")
        plt.xlabel("X-axis (pixels)")
        plt.ylabel("Y-axis (pixels)")

        # Set fixed axis limits
        plt.xlim(40, 62)
        plt.ylim(28, 42)

        # Display or save the plot
        plt.show()
        # Optionally, save the plot:
        # plt.savefig(f"velocity_timestep_{timestep}.png")
        plt.close()

# Path to the HDF5 file
h5_file_path = r"Your .h5 file path"

# Run the function to visualize the elliptical ROI for all timesteps
visualize_ellipse_for_all_timesteps(h5_file_path)
```
