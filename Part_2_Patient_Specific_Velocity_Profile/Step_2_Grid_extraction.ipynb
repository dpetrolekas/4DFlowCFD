```python
#Grid data extraction (one timestep)
import numpy as np
import h5py

# File paths
h5_file_path = r"Your Path\mri_data.h5"
output_file_path = r"Your Path\Support_files\5_coor_vel_grid.txt"

# Load the HDF5 file
with h5py.File(h5_file_path, "r") as f:
    # Select timestep 
    timestep = 4

    # Load velocity components (Shape: [sagittal (y), axial (z), coronal (x)])
    u = f["u"][timestep, :, :, :]
    v = f["v"][timestep, :, :, :]
    w = f["w"][timestep, :, :, :]

# Get dimensions
num_y, num_z, num_x = u.shape

# Create an array to store coordinates and velocities
coor_vel_data = []

# Loop through the axial slices (z index)
for z in range(num_z):
    for y in range(num_y):
        for x in range(num_x):
            coor_vel_data.append([x, y, z, u[y, z, x], v[y, z, x], w[y, z, x]])

# Convert to NumPy array
coor_vel_array = np.array(coor_vel_data)

# Save to TXT file in the desired directory
header = "X\tY\tZ\tU\tV\tW"
np.savetxt(output_file_path, coor_vel_array, fmt="%.6f", delimiter="\t", header=header, comments="")

print(f"Velocity grid successfully saved to:\n{output_file_path}")

# Print some sample values
print("Sample coordinate-velocity data:")
print(coor_vel_array[:5])
```
