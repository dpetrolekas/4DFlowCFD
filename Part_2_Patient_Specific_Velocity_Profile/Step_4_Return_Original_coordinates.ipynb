```python
#Step_1:descale
import numpy as np

# File paths
input_file_path = r"Your Path\Support_files\10_Coor_Cap_interpolated_timesteps.txt"
output_file_path = r"Your Path\Support_files\11_Cap_interpolated_timesteps_step_1.txt"
mri_file_path = r"Your Path\Support_files\3_cap_coor_mri.txt"

# Load MRI coordinates to extract min/max values
mri_data = np.loadtxt(mri_file_path, delimiter="\t", skiprows=1)  # Skip header

# Compute min/max for X and Y
x_mri_min, x_mri_max = np.min(mri_data[:, 0]), np.max(mri_data[:, 0])
y_mri_min, y_mri_max = np.min(mri_data[:, 1]), np.max(mri_data[:, 1])

# Transformation parameters
desired_xmin, desired_xmax = 41, 62
desired_ymin, desired_ymax = 29, 41

scale_x = (desired_xmax - desired_xmin) / (x_mri_max - x_mri_min)
scale_y = (desired_ymax - desired_ymin) / (y_mri_max - y_mri_min)
scale_z = 191 / 29.85  # Z-axis scaling factor

# Read the input file
with open(input_file_path, 'r') as f:
    lines = f.readlines()

# Initialize storage for transformed data
transformed_data = []
current_timestep = None

# Process each line
for line in lines:
    line = line.strip()
    
    if line.startswith("aligned_comp_"):  # Timestep header
        current_timestep = line
        transformed_data.append(f"{current_timestep}\n")
    
    elif line == "x, y, z, u, v, w":  # Header line
        transformed_data.append("x, y, z, u, v, w\n")
    
    elif line:  # Process numerical data
        values = list(map(float, line.split(',')))
        x_scaled, y_scaled, z_scaled, u, v, w = values
        
        # Apply inverse transformation
        x_mri = ((x_scaled - desired_xmin) / scale_x) + x_mri_min
        y_mri = ((y_scaled - desired_ymin) / scale_y) + y_mri_min
        z_mri = z_scaled / scale_z  # Z transformation
        
        # Store transformed data
        transformed_data.append(f"{x_mri:.6f}, {y_mri:.6f}, {z_mri:.6f}, {u:.6f}, {v:.6f}, {w:.6f}\n")
    
    else:  # Blank line (keeps formatting)
        transformed_data.append("\n")

# Save the transformed data
with open(output_file_path, 'w') as f:
    f.writelines(transformed_data)

print("\n✅ Transformation complete! Saved to:", output_file_path)
print(f"x_mri_min: {x_mri_min}, x_mri_max: {x_mri_max}")
print(f"y_mri_min: {y_mri_min}, y_mri_max: {y_mri_max}")
```
